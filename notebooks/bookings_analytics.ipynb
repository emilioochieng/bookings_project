```python
import pandas as pd
import numpy as np
from pathlib import Path

RAW = Path("../data/raw")

# 1. load and combine the three booking files
df_2023 = pd.read_csv(RAW / "Bookings_2023.csv")
df_2024 = pd.read_csv(RAW / "Bookings_2024.csv")
df_2025 = pd.read_csv(RAW / "Bookings_2025.csv")
df = pd.concat([df_2023, df_2024, df_2025], ignore_index=True)
print(df.shape)                                   # (13927, 12)

# 2. clean: duplicates, status spellings, dates
df = df.drop_duplicates(subset="BookingID", keep="first")
df["Status"] = df["Status"].str.strip().str.title().replace({"Canceled": "Cancelled"})
df["BookingDate"] = pd.to_datetime(df["BookingDate"])
df["BookingYear"] = df["BookingDate"].dt.year

# 3. calculated columns
df["Travellers"] = df["Adults"] + df["Children"]
df["Grossvalue"] = (df["Adults"] + 0.5 * df["Children"]) * df["UnitPriceUSD"]
df["Discount"] = df["Grossvalue"] * df["DiscountPct"].fillna(0)
df["NetValue"] = df["Grossvalue"] - df["Discount"]
print(df.shape)                                   # (13890, 17)

# 4. load the lookup tables
packages = pd.read_csv(RAW / "Packages.csv")
agents = pd.read_csv(RAW / "Agents.csv")
```

```python
live = df[df["Status"] != "Cancelled"].copy()
live["AgentID"] = live["AgentID"].fillna("AG00")

live = live.merge(packages, on="PackageID", how="left")
live = live.merge(agents[["AgentID", "AgentName", "Channel", "CommissionPct"]], on="AgentID", how="left")

live["GrossProfit"] = (live["NetValue"]
                       - live["Travellers"] * live["CostPerPaxUSD"]
                       - live["NetValue"] * live["CommissionPct"])

package_perf = (
    live.groupby(["PackageID", "PackageName"]).agg(
        Bookings=("BookingID", "nunique"),
        Revenue=("NetValue", "sum"),
        Profit=("GrossProfit", "sum"),
    ).reset_index()
)
package_perf["Margin_%"] = (package_perf["Profit"] / package_perf["Revenue"] * 100).round(1)

print(len(live))     # 12,578
```

```python
best_packages = package_perf.merge(
    packages[["PackageID", "Category", "DurationDays", "BasePriceUSD"]],
    on="PackageID", how="left"
)
best_packages = best_packages.sort_values("Revenue", ascending=False).reset_index(drop=True)

top3 = best_packages.head(3)
top3.round(1)
```

```python
adult_value = (live["Adults"] * live["UnitPriceUSD"]).sum()
child_value = (0.5 * live["Children"] * live["UnitPriceUSD"]).sum()
gross = live["Grossvalue"].sum()
discount = live["Discount"].sum()
net = live["NetValue"].sum()

components = pd.DataFrame({
    "Component": ["Adults", "Children (half price)", "Gross value", "Discounts", "Net value"],
    "USD": [adult_value, child_value, gross, -discount, net],
})
components["Share_of_gross_%"] = (components["USD"] / gross * 100).round(1)
components.round(1)
```

```python
by_category = live.groupby("Category").agg(
    Bookings=("BookingID", "nunique"),
    NetValue=("NetValue", "sum"),
).reset_index()
by_category["Share_%"] = (by_category["NetValue"] / by_category["NetValue"].sum() * 100).round(1)
by_category.sort_values("NetValue", ascending=False)
```

```python
by_channel = live.groupby("Channel").agg(
    Bookings=("BookingID", "nunique"),
    NetValue=("NetValue", "sum"),
).reset_index()
by_channel["Share_%"] = (by_channel["NetValue"] / by_channel["NetValue"].sum() * 100).round(1)
by_channel.sort_values("NetValue", ascending=False)
```

```python
df = pd.concat([pd.read_csv(RAW / f"Bookings_{y}.csv") for y in (2023, 2024, 2025)], ignore_index=True)

df = df.drop_duplicates(subset="BookingID", keep="first")
df["Status"] = df["Status"].str.strip().str.title().replace({"Canceled": "Cancelled"})
df["AgentID"] = df["AgentID"].fillna("AG00")
df["DiscountPct"] = df["DiscountPct"].fillna(0)

df["BookingDate"] = pd.to_datetime(df["BookingDate"])
df["TravelStartDate"] = pd.to_datetime(df["TravelStartDate"])
df["BookingYear"] = df["BookingDate"].dt.year
df["BookingMonth"] = df["BookingDate"].dt.to_period("M").astype(str)

df = df.merge(packages[["PackageID", "PackageName", "Category", "DurationDays", "PrimaryParkID", "CostPerPaxUSD"]],
              on="PackageID", how="left")
df = df.merge(agents[["AgentID", "AgentName", "Channel", "CommissionPct"]], on="AgentID", how="left")

df["Travellers"] = df["Adults"] + df["Children"]
df["Grossvalue"] = (df["Adults"] + 0.5 * df["Children"]) * df["UnitPriceUSD"]
df["Discount"] = df["Grossvalue"] * df["DiscountPct"]
df["NetValue"] = df["Grossvalue"] - df["Discount"]
df["Cost"] = df["Travellers"] * df["CostPerPaxUSD"]
df["Commission"] = df["NetValue"] * df["CommissionPct"]
df["GrossProfit"] = df["NetValue"] - df["Cost"] - df["Commission"]
df["IsActive"] = (df["Status"] != "Cancelled").astype(int)

final_cols = ["BookingID", "BookingDate", "BookingYear", "BookingMonth", "TravelStartDate", "CustomerID",
              "PackageID", "PackageName", "Category", "DurationDays", "PrimaryParkID", "AgentID", "AgentName",
              "Channel", "Adults", "Children", "Travellers", "UnitPriceUSD", "DiscountPct", "Grossvalue",
              "Discount", "NetValue", "CostPerPaxUSD", "Cost", "CommissionPct", "Commission", "GrossProfit",
              "Status", "IsActive", "PaymentMethod"]

final = df[final_cols].sort_values(["BookingDate", "BookingID"]).reset_index(drop=True)

OUT = Path("../data/processed")
OUT.mkdir(parents=True, exist_ok=True)
final.to_csv(OUT / "Bookings_clean_combined.csv", index=False)
```
